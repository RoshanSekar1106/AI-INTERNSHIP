```python
transactions=[]
for i in range(0,len(df)):
    for j in range(0,len(df.columns)):
      transactions.append(str(df.values[i,j]))
```

```python
from apyori import apriori

rules=apriori(transactions=transactions,min_support=0.003,min_confidence=0.2,min_lift=3,min_length=2,max_length=2)

result = list(rules)
```

```python
result
```

```python
data = []

for item in result:
    support = item.support

    for stat in item.ordered_statistics:
        lhs = ', '.join(list(stat.items_base))
        rhs = ', '.join(list(stat.items_add))
        confidence = stat.confidence
        lift = stat.lift

        data.append([lhs, rhs, support, confidence, lift])

rules_df = pd.DataFrame(data,
    columns=['LHS', 'RHS', 'Support', 'Confidence', 'Lift']
)

print(rules_df.head())
```
